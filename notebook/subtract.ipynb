```python
# Define inputs
a = 10
b = 5

def subtract(a, b):
    """Subtracts two numbers."""
    return a - b

result = subtract(a, b)
print(result)
print(f"{a} - {b} = {result}")

def distance_from_zero(num):
    """Calculates the distance of a number from zero."""
    return abs(num)

distance = distance_from_zero(a)
print(f"Distance of {a} from zero is {distance}")
```
